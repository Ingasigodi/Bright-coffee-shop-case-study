```sql
---Checking whole table
SELECT *
FROM workspace.brightcoffeshop.brightcoffeeshop;

---checking how many rows and no duplicates
SELECT COUNT(*) AS num_of_rows,
       COUNT(DISTINCT transaction_id) AS user_id
FROM workspace.brightcoffeshop.brightcoffeeshop;

--- checking for nulls in the data set
SELECT COUNT(*) AS NULL_ROWS
FROM workspace.brightcoffeshop.brightcoffeeshop
WHERE transaction_id IS NULL;

---checking the start date and end date of the dataset
SELECT MIN(transaction_date) AS earliest_date,
        MAX(transaction_date) AS latest_date
FROM workspace.brightcoffeshop.brightcoffeeshop;


---calculating revenue for bright coffee shop
SELECT ROUND(SUM(transaction_qty * TRY_CAST(REPLACE(unit_price,',','.')AS DOUBLE)),2) AS TOTAL_Revenue
FROM workspace.brightcoffeshop.brightcoffeeshop;

---calculating the distinct branches
SELECT DISTINCT store_location,
                store_id

FROM workspace.brightcoffeshop.brightcoffeeshop;

---Distinct product category
SELECT DISTINCT product_category
FROM workspace.brightcoffeshop.brightcoffeeshop;

---Checking the lowest and highest prices in the unit_price column
SELECT MIN(unit_price) AS Lowest_unit_price,
       MAX(unit_price) AS highest_unit_price
FROM workspace.brightcoffeshop.brightcoffeeshop;     

---Checking the total revenue by store location
SELECT  DISTINCT store_location,
        ROUND(SUM(transaction_qty * TRY_CAST(REPLACE(unit_price,',','.')AS DOUBLE)),2)
FROM workspace.brightcoffeshop.brightcoffeeshop
GROUP BY store_location;

---Number of transactions
SELECT COUNT (DISTINCT transaction_id) AS number_of_transactions
FROM workspace.brightcoffeshop.brightcoffeeshop;

---AVG number of transactions
SELECT ROUND(AVG(transaction_qty * TRY_CAST(REPLACE(unit_price,',','.')AS DOUBLE)),2) AS AVG_transactions
FROM workspace.brightcoffeshop.brightcoffeeshop;

---Big cleaned query
SELECT transaction_id,
       transaction_date,
       transaction_time,
       transaction_qty,
       store_id,
       store_location,
       product_id,
       unit_price,
       product_category,
       product_type,
       product_detail,
       DAYNAME(transaction_date) AS date_name,
       MONTHNAME(transaction_date) AS month_name,
       HOUR(transaction_time) AS transaction_hour,
       CASE   
        WHEN HOUR(transaction_time) BETWEEN 0 AND 2 THEN 'wee hours' 
        WHEN HOUR(transaction_time) BETWEEN 3 AND 5 THEN 'ungodly hours' 
        WHEN HOUR(transaction_time) BETWEEN 6 AND 8 THEN 'rush hour' 
        WHEN HOUR(transaction_time) BETWEEN 9 AND 11 THEN 'mid morning' 
        WHEN HOUR(transaction_time) BETWEEN 12 AND 14 THEN 'Afternoon' 
        WHEN HOUR(transaction_time) BETWEEN 15 AND 17 THEN 'Late afternoon' 
        WHEN HOUR(transaction_time) BETWEEN 18 AND 20 THEN 'evening' 
        ELSE 'night'
       END AS transaction_time_bucket
FROM workspace.brightcoffeshop.brightcoffeeshop;
```
